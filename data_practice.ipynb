```python
import requests
```

```python
URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)
if res.status_code == 200:
    print(res.headers)

with open("scraped_data/data1.html", "w") as f:
    f.write(res.text)
     
```

```python
from bs4 import BeautifulSoup

with open("scraped_data/data1.html","r")as f:
    html_content = f.read()
soup = BeautifulSoup(html_content,"lxml")
soup
all_h3 = soup.find_all("h3")
all_contries = []
for h3 in all_h3:
    name = h3.get_text(strip=True)

div = h3.find_next("div")
population = div.select("span.country-population")
for p in population:
    print(p.get_text(strip=True))
all_contries.append([name,population])

```

```python
all_contries
import pandas as pd
df = pd.DataFrame(all_contries, columns=["Name","population"])
df
```

```python
df.to_csv("cleaned_data/data.csv",index=False)
```

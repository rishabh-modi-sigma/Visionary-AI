```python
import requests
from bs4 import BeautifulSoup
import os
```

```python
os.makedirs("Quotes_all",exist_ok=True)
page_count = 1
while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text,"lxml")
    quotes = soup.select("div .quote")

    if not quotes:
        print("No valid pages anmore")
        break
    with open(f"Quotes_all/quotes{page_count}.html","w")as f:
        f.write(res.text)
        
    print(f"downlod from page{page_count}")
    page_count += 1
```

```python
from bs4 import BeautifulSoup

with open("Quotes_all/quotes1.html","r") as f:
    html_count = f.read()

soup = BeautifulSoup(html_count,"lxml")

author_count = {}   # dictionary
all_quotes = soup.select(".quote")

for q in all_quotes:
    all_tags = []

    for tag in q.select(".tags .tag"):
        all_tags.append(tag.get_text(strip=True))

    # if "life" in all_tags:
        author = q.select_one(".author").get_text(strip=True)

        if author in author_count:
            author_count[author] += 1
        else:
            author_count[author] = 1

# Output
for author, count in author_count.items():
    print(f"{author} : {count} quotes")

```

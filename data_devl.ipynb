```python
import requests
from bs4 import BeautifulSoup
import os

# Folder create karo agar nahi hai
os.makedirs("Quotes_data", exist_ok=True)

page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    if res.status_code != 200:
        print("No more pages available")
        break

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select(".quote")

    if not quotes:
        print("No valid pages anymore")
        break

    with open(f"Quotes_data/quotes{page_count}.html", "w", encoding="utf-8") as f:
        f.write(res.text)

    print(f"Downloaded data from page {page_count}")
    page_count += 1
```

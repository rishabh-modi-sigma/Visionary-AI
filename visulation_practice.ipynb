```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "student":["Amit","Riya","Sohan","Neha","Kunal","Pooja","Ankit","Simran"],
    "Math":[78,85,67,90,72,88,60,95],
    "Science":[82,88,70,92,75,90,65,98],
    "English":[75,80,72,85,70,82,68,90],
    "Sex":["M","F","M","F","M","F","M","F"]
}
df = pd.DataFrame(data)
df
```

```python
sns.barplot(
    data=df,
    x="student",
    y="Math",
    hue="Sex"
)
plt.title("Student Performance Data")
plt.tight_layout()
```

```python
data = {
     "Math":[78,85,67,90,72,88,60,95],
    "Science":[82,88,70,92,75,90,65,98],
    "English":[75,80,72,85,70,82,68,90]
}

df = pd.DataFrame(data)
sns.boxplot(
    data = df
)
plt.title("Box Plot of subject Marks")
plt.ylabel("Marks")
plt.show()
```

```python
data = {
   "Math":[78,85,67,90,72,88,60,95],
   "Science":[82,88,70,92,75,90,65,98]
}

df = pd.DataFrame(data)

sns.scatterplot(
    data = df,
    x = "Math",
    y = "Science"
 
)
plt.title("Math vs Science Marks Plot")
plt.xlabel("Math")
plt.ylabel("Science")
plt.show()
```

```python
data = {
    "Math":[78,85,67,90,72,88,60,95],
    "Science":[82,88,70,92,75,90,65,98],
    "English":[75,80,72,85,70,82,68,90]
}

df = pd.DataFrame(data)
# Correlation matrix
corr = df.corr()

# Heatmap
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Subjects")
plt.show()
```

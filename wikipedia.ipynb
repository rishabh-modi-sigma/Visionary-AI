```python
import os
import requests
from bs4 import BeautifulSoup

os.makedirs("wikipedia", exist_ok=True)

URL = "https://en.wikipedia.org/wiki/Manali,_Himachal_Pradesh"

headers = {
    "User-Agent": "Mozilla/5.0 (Educational Web Scraping Bot)"
}

res = requests.get(URL, headers=headers)

if res.status_code == 200:
    soup = BeautifulSoup(res.text, "lxml")

    with open("wikipedia/manali_full_page.html", "w", encoding="utf-8") as f:
        f.write(res.text)

    print("Page successfully saved")
else:
    print("Page not available")
```

```python
with open("wikipedia/manali_full_page.html","r") as f:
    html_data = f.read()

soup = BeautifulSoup(html_data,"lxml")
hading = soup.select_one("#firstHeading").get_text(strip=True)

hading_data = []
hading_data.append(hading)
print(hading_data)
```

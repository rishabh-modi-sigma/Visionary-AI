```python
import pandas as pd

# Load dataframe again (important)
df = pd.read_csv("raw_data.csv")

# remove extra spaces in column names
df.columns = df.columns.str.strip()

# tax column
df["tax"] = df["income"].apply(
    lambda x: "20%" if x >= 50000 else "10%"
)

# gender mapping
gender_map = {"Male":"M","Female":"F","Unknown":"U"}
df["gender"] = df["gender"].map(gender_map)

# assign
df = df.assign(new_income = df["income"] * 1.1) 

#replace
df["country"] = df["country"].replace("USA","US")
df
```

```python
df.columns=["Id","Name","Age","Country","Gender","Income","Tax","New_Index"]
df
df.rename(columns={"Gender":"Gender"})
df.rename(index={2:"Two"})

df.sort_values("Income")
df.sort_values("Income",ascending=False)
df = df.fillna(50)
sorted_df = df.sort_values(["Income","Age"])
df
sorted_df.sort_index()
sorted_df.reset_index(drop=True)

#Ranking
sorted_df["Ranking"] = sorted_df["Income"].rank(ascending=False,method="max")
sorted_df
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
page_count = 1
while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soop = BeautifulSoup(res.text,"lxml")
    quotes = soop.select("div .quote")

    if not quotes:
        print("no valid page anmore")
        break

    with open(f"scrapeds_data/quotes{page_count}.html","w") as f:
        f.write(res.text)

    print(f"downloded data from page {page_count}")
    page_count = page_count+1
```

```python
with open(f"scraped_data/quotes1.html","r") as f:
     html_content = f.read()
all_quotes_data = []
qutes_data = []

soop = BeautifulSoup(html_content,"lxml")
all_quotes = soop.select(".quote")

for q in all_quotes:
    tags = []

for tag in q.select(".tags .tag"):
  tags.append(tag.get_text(strip=True))

for q in all_quotes:
    text = q.select_one(".text").get_text(strip=True)
    authors = q.select_one(".author").get_text(strip=True)
    qutes_data.append([text , authors])

    all_quotes_data.append({"qutoes":text,
                        "authors":authors,
                        "tags":tags})

all_quotes_data
```
